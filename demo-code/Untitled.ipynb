```python
# Import Statements go here
import numpy as np
import numpy.linalg as npla
from scipy import sparse

def pagerank2(E, return_vector = False, max_iters = 1000, tolerance = 1e-6, m = 0.15):
    nrow,ncol = E.shape
    assert nrow==ncol, 'E must be square'
    assert np.max(E)==1 and np.sum(E)==np.count_nonzero(E), 'E must contain only zeros and ones'

    if type(E) is not sparse.csr_matrix:
       print('Warning, converting input from type', type(E), 'to sparse matrix.')
       E = sparse.csr_matrix(E)
    
    print("E Matrix:", E.todense())
 
    outdegree = np.zeros(nrow)
    for i in range(E.size):
       outdegree[E.indices[i]]+=1
    

    e = np.ones(nrow)
    v = e/sum(e)
    
    print('original v:',v)

    there = np.where(outdegree==0)

    for iteration in range(max_iters):
        print("_v", v)
        oldv = v
        SV = np.sum(v)/nrow

        #
        #EV[np.where(outdegree!=0)] = EV[np.where(outdegree!=0)]/outdegree
        #EV = E@v
        #EV = EV / sum(EV)
 
        
        w = np.zeros(nrow)
        non_dangling = outdegree != 0
        w[non_dangling] = v[non_dangling] / outdegree[non_dangling]
        EV = E @ w
        print("EV", EV)
        
        FV = np.ones(nrow)*np.sum(v[there])
        FV[there] -= v[there]

        FV = FV/(nrow-1)

        MV = (1-m)*(EV+FV) + (m*SV)
        print("MV", MV)

  
        v = MV
        eigval = npla.norm(v)
        v = v / eigval

        if npla.norm(v-oldv) < tolerance:
           break
        print("-------------")

    if npla.norm(v-oldv) < tolerance:
       print('Dominant eigenvalue is %f after %d iterations.\n' % (eigval, iteration+1))
    else:
       print('Did not converge to tolerance %e after %d iterations.\n' % (tolerance, max_iters))

    #assert np.all(v>0) or np.all(v<0), 'Error: eigenvector is not all > 0 or < 0'
    #print('eigenvector:',v)
    #print(npla.norm(v-oldv))
    vector = np.abs(v)
    #print('outdegree:',outdegree)

    ranking = np.argsort(vector)[::-1]
    if return_vector:
      return ranking,vector
    else:
      return ranking  


# test function

E = np.load('PageRankEG1.npy')
r, v = pagerank2(E, return_vector = True, max_iters = 1000, tolerance = 1e-8)
print('r =', r)
print('v =', v)
```
