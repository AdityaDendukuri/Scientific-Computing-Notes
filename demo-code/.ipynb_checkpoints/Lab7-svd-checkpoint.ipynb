# Lab 7: Rank and Singular Value Decomposition

We have previously encountered the notion of **singular** matrices. For a singular matrix $A$, we've learned that:

1. $A^{-1}$ does not exist.
2. The equation $Ax = b$ does not have a unique solution.
3. The condition number $\text{cond}(A)$ is infinite.
4. The determinant $\text{det}(A)$ equals $0$.

**Singular Value Decomposition** (SVD) provides a comprehensive framework that ties these concepts together. Before delving into SVD, let's recap the concepts of **rank** and <b>outer product</b>.

### Rank

The **rank** of a matrix $A$ has several definitions, but for our purposes, we'll adopt the following:

> The **rank** of a matrix $A$ is the number of linearly independent rows (or columns) it possesses.

**Example** - Consider the matrix with three rows $r_1$, $r_2$, and $r_3$ (dashed lines represent the rest of the row):
$$
A = \begin{bmatrix} -- & r_1 & -- \\ 
                    -- & r_2 & -- \\
                    -- & r_3 & -- \end{bmatrix} =
    \begin{bmatrix} 1 & 1 & 0 \\ 
                    0 & 0 & 1 \\
                    2 & 2 & 0 \end{bmatrix}.
$$
1. We observe two linearly independent rows ($r_1$ and $r_2$). Thus, $\text{rank}(A) = 2$.
2. $r_3$ is not linearly independent, as it can be expressed as a linear combination of $r_1$ and $r_2$ (specifically, $r_3 = 2*r_1 + 0*r_2$).

```python
A = np.array( [[1, 1, 0],
               [0, 0, 1],
               [2, 2, 0]] )
np.linalg.matrix_rank(A)
```

### Outer Product

>2. Inner product: $u^{T} v: (1, m)\cdot(m, 1) = 1$ dimensional scalar. Also called dot product.

```python
a = np.random.rand(5, 1)
b = np.random.rand(5, 1)
outer_prod = a @ b.T
inner_prod = b.T @ a 

#Note: if a is a 1D ndarray, then you have to use np.outer as you as you cannot transpose 
a     = np.random.rand(5)
b     = np.random.rand(5)
outer = np.outer(a, b)
outer = np.inner(a, b)
```

### Singular Value Decomposition (SVD)

The Singular Value Decomposition (SVD) is a mathematical technique that breaks down a matrix $A$ into three component matrices, each providing valuable insight into the matrix's structure and geometry.

>$$ A = U \Sigma V^T $$
>
>1. $U$ is an $m \times m$ orthogonal matrix whose columns are the **left singular vectors**.
>2. $\Sigma$ is an $m \times n$ diagonal matrix containing the singular values $\sigma_1 \geq \sigma_2 \geq \dots \geq 0$.
>3. $V$ is an $n \times n$ orthogonal matrix whose columns are the **right singular vectors**.

In NumPy, `np.linalg.svd(A)` returns `U, S, Vt`, where `Vt` is $V^T$, not $V$.

```python
# Generate a random full rank matrix 
A = np.random.rand(10, 10)
m, n = A.shape

# compute SVD of A
U, S, Vt = np.linalg.svd(A)

# reconstruct A
A_ = U @ np.diag(S) @ Vt

# verify properties 
assert np.allclose(A, A_)
assert np.allclose(U.T @ U, np.eye(m))
assert np.allclose(Vt @ Vt.T, np.eye(n))
```


**Geometrical Interpretation of SVD:**

The SVD says that $A$ acts in three clean steps:

$$
x \xrightarrow{V^T} \text{write } x \text{ in the right-singular-vector basis}
\xrightarrow{\Sigma} \text{scale each coordinate}
\xrightarrow{U} \text{rotate into the output space}.
$$

The key relation is

$$
A v_i = \sigma_i u_i.
$$

So $v_i$ is an input direction, $u_i$ is the corresponding output direction, and $\sigma_i$ is the amount of stretching in that direction. The unit circle or sphere is mapped to an ellipse or ellipsoid whose semi-axis lengths are the singular values.

**Rank and zero singular values:**

The rank is the number of nonzero singular values:

$$
\operatorname{rank}(A) = \#\{i : \sigma_i > 0\}.
$$

If $\sigma_i = 0$, then

$$
A v_i = 0,
$$

so that input direction is completely collapsed. In other words, the corresponding right singular vector lies in the nullspace of $A$.

**Relating Condition Number to SVD:**

Since singular values measure directional scaling, the spectral condition number is

$$
\kappa_2(A) = \frac{\text{largest stretch}}{\text{smallest stretch}}
= \frac{\sigma_{\max}}{\sigma_{\min}}.
$$

If $\sigma_{\min}$ is very small, then $A$ nearly collapses some direction, so solving $Ax=b$ is sensitive to small perturbations. If $\sigma_{\min}=0$, the matrix is singular and $\kappa_2(A)=\infty$.

**Four fundamental subspaces from SVD:**

If $A \in \mathbb{R}^{m \times n}$ has rank $r$, then the SVD gives bases for the four fundamental subspaces:

$$
\operatorname{Col}(A) = \operatorname{span}\{u_1,\dots,u_r\},
$$

$$
\operatorname{Null}(A^T) = \operatorname{span}\{u_{r+1},\dots,u_m\},
$$

$$
\operatorname{Row}(A) = \operatorname{span}\{v_1,\dots,v_r\},
$$

$$
\operatorname{Null}(A) = \operatorname{span}\{v_{r+1},\dots,v_n\}.
$$

Thus, SVD separates the input space into directions that survive and directions that vanish, and separates the output space into directions that can be reached and directions that cannot.

```python
A = np.random.rand(10, 10)
U, S, Vt = np.linalg.svd(A) 

cond_num_numpy = np.linalg.cond(A) 
cond_num_svd   = S[0] / S[-1]

assert np.isclose(cond_num_svd, cond_num_numpy)
```

### Visualizing SVD Geometry

The next cell shows the geometric picture directly. A full-rank matrix maps the unit circle to an ellipse. A rank-deficient matrix collapses the unit circle to a lower-dimensional set.

```python
def draw_vector(ax, v, label):
    ax.arrow(0, 0, v[0], v[1], head_width=0.05, length_includes_head=True)
    ax.text(1.08*v[0], 1.08*v[1], label, fontsize=12)

# Full-rank example: circle -> ellipse
A_full = np.array([[2.0, 1.2],
                   [0.6, 0.7]])

U_full, S_full, Vt_full = np.linalg.svd(A_full)
V_full = Vt_full.T

theta = np.linspace(0, 2*np.pi, 400)
circle = np.vstack((np.cos(theta), np.sin(theta)))
ellipse = A_full @ circle

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].plot(circle[0], circle[1])
draw_vector(axs[0], V_full[:, 0], r"$v_1$")
draw_vector(axs[0], V_full[:, 1], r"$v_2$")
axs[0].set_title("Input directions")
axs[0].set_aspect("equal")
axs[0].grid(True)
axs[0].axhline(0, linewidth=0.8)
axs[0].axvline(0, linewidth=0.8)

axs[1].plot(ellipse[0], ellipse[1])
draw_vector(axs[1], S_full[0] * U_full[:, 0], r"$\sigma_1 u_1$")
draw_vector(axs[1], S_full[1] * U_full[:, 1], r"$\sigma_2 u_2$")
axs[1].set_title("Output ellipse")
axs[1].set_aspect("equal")
axs[1].grid(True)
axs[1].axhline(0, linewidth=0.8)
axs[1].axvline(0, linewidth=0.8)

plt.tight_layout()
plt.show()

print("Singular values:", S_full)
print("Condition number:", S_full[0] / S_full[-1])
```

```python
# Rank-deficient example: one direction is completely collapsed
A_rank_def = np.array([[2.0, 0.0],
                       [0.0, 0.0]])

U_rd, S_rd, Vt_rd = np.linalg.svd(A_rank_def)
V_rd = Vt_rd.T

theta = np.linspace(0, 2*np.pi, 400)
circle = np.vstack((np.cos(theta), np.sin(theta)))
image = A_rank_def @ circle

plt.figure(figsize=(5, 4))
plt.plot(circle[0], circle[1], label="unit circle")
plt.plot(image[0], image[1], label="image under A")
plt.title("Rank-deficient map: circle collapses to a line segment")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()

print("Singular values:", S_rd)
print("Numerical rank:", np.linalg.matrix_rank(A_rank_def))
print("Nullspace direction:", V_rd[:, -1])
print("A @ nullspace_direction:", A_rank_def @ V_rd[:, -1])
```

**Relating SVD to Frobenius norm**

The Frobenius norm of a matrix is defind as -

$$
\|A\|_F = \sqrt{\sum_{i=1}^{m} \sum_{j=1}^{n} |a_{ij}|^2}
$$


If we plug in in the SVD decomposition and simplify we get


$$
\|A\|_F = \sqrt{\sum_{i=1}^{r} \sigma_i^2}
$$

Side note: Similar to vectors, we can compute a <i>residual</i> matrix to compare two matrices (say $A$ and $A'$) and compute the Frobenius norm as 

$$
||R||_F = \|A - A'\|_F.
$$

## Rank Reduction using Singular Value Decomposition (SVD)

The Singular Value Decomposition (SVD) breaks down a matrix $A$ into three components: the left singular vectors ($U$), the singular values ($\Sigma$), and the right singular vectors ($V^\text{T}$). These components are arranged as follows:

$$
A = U \Sigma V^\text{T} = 
\begin{bmatrix}
| & & | & |& & | \\
| & & | & |& & | \\
| & & | & |& & | \\
u_{1} & \dots & u_{r} & u_{r+1} & \dots & u_{m} \\
| & & | & |& & | \\
| & & | & |& & | \\
| & & | & |& & | \\
\end{bmatrix}
\begin{bmatrix}
\sigma_{1} & & & & & \\
& \ddots & & & & \\
& & \sigma_{r} & & & \\
& & & 0 & & \\
& & & & \ddots & \\
& & & & & 0 \\
\end{bmatrix}
\begin{bmatrix}
- & - & v_{1}^{\text{T}} & - & - \\
& & \vdots & & \\
- & - & v_{r}^{\text{T}} & - & - \\
- & - & v_{r+1}^{\text{T}} & - & - \\
& & \vdots & & \\
- & - & v_{n}^{\text{T}} & - & - \\
\end{bmatrix}
$$

Here:
1. $r$ is the rank of the matrix $A$.
2. For non-singular square matrices, $\Sigma$ has non-zero singular values.
3. For singular or non-square matrices, $\Sigma$ has $r$ non-zero singular values along the main diagonal.

By simplifying the matrix product, $A$ we get:

$$ A = \sum_{i=1}^{r} \sigma_i u_i v_i^{\text{T}} $$

If we truncate this sum to include only the first $m$ terms, the resulting matrix $A'$ has a rank of $m$, reducing the complexity of $A$:

$$ A' = \sum_{i=1}^{m} \sigma_i u_i v_i^{\text{T}} $$

By selecting $m < \text{rank}(A)$, we retain the most significant components of $A$, effectively reducing its rank and complexity. This process is known as rank reduction using SVD.

```python
# Rank reduction (change r value to reduce it to a rank you want) using the loop summation
A_approx = np.zeros(shape=A.shape)
r = 9
for i in range(r):
    A_approx += S[i] * np.outer(U[:, i], Vt[i, :])
    
print(f"Frobenius error for rank {r} reduction: {np.linalg.norm(A - A_approx, ord='fro')}")
```

> **Note:** We can also represent the truncation in the matrix product form using array slicing, 

$$ A' = U[:, :r] \, \Sigma[:r, :r] \, V^{T}[:r, :] $$

```python
# You can also do the above summation operation using slicing
A_approx = U[:, :r] @ np.diag(S[:r]) @ Vt[:r, :]

print(f"Frobenius error for rank {r} reduction: {np.linalg.norm(A - A_approx, ord='fro')}")
```

## Rank Reduction Example: Low rank approximations of images

An image is represented as a 3D matrix, with each element denoting a pixel's RGB color intensity (ranging from 0 to 255). To set a pixel's intensity, we use the notation: $$X[i, j, :] = [r, g, b]$$

In this demonstration, images are converted to grayscale, resulting in a 2D matrix where each element represents the average of the RGB values: $$X_{\text{grey}}[i, j] = \frac{r + g + b}{3}$$

```python
X = plt.imread("henley.jpg")
X = X.mean(axis=2) # take the mean of the rgb intensities (the third dimension)
#X = 0.299*X[:, :, 0] + 0.587*X[:, :, 1] + 0.114*X[:, :, 2]

plt.title(f'Full Rank :{np.linalg.matrix_rank(X)}')
plt.imshow(X, cmap='gray')
plt.show()
```

```python
U, S, Vt = np.linalg.svd(X, full_matrices=False)
X_approx = np.zeros(shape=X.shape) 

# do rank reductions for the following ranks 
ranks = [5, 10, 200, 700]
# ranks = np.arange(1, 500)

# perform rank reductions for ranks defined above and compute reconstruction error
X_lowrank = []
errors = []
for r in ranks:
    X_approx = U[:, :r] @ np.diag(S[:r]) @ Vt[:r, :]
    errors.append(np.linalg.norm(X_approx - X, ord="fro") / np.linalg.norm(X, ord="fro"))
    X_lowrank.append(X_approx)

# plot values
plt.plot(ranks, errors, '-o')
plt.xlabel("reduced rank $r$")
plt.ylabel(r"Reconstruction error $\|X - U_r \Sigma_r V_r^T\|_F / \|X\|_F$")
plt.show()
```

```python
# compute error and display approximations
fig, axs = plt.subplots(1, len(ranks), figsize=(10, 10))

for i in range(len(ranks)):    
    axs[i].set_title(f"rank: {ranks[i]} | err: {errors[i]:.3f}")
    axs[i].imshow(X_lowrank[i], cmap='gray')
    
plt.tight_layout()
plt.show()
```
