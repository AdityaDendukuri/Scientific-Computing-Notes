```python
import numpy as np
import numpy.linalg as npla
import matplotlib.pyplot as plt
from sklearn import datasets
```

# PCA and PageRank

Two applications that sit on top of the same core tool: eigendecomposition / SVD.

---
## Part 1 — Principal Component Analysis (PCA)

### Covariance matrix

Given a data matrix $D \in \mathbb{R}^{n \times p}$ (n samples, p features), center it:

$$Z = D - \mathbf{1}\mu^T, \qquad \mu_j = \frac{1}{n}\sum_{i=1}^n D_{ij}.$$

The **covariance matrix** is

>$$C = \frac{1}{n} Z^T Z \in \mathbb{R}^{p \times p}.$$

$C$ is **symmetric positive semi-definite** (SPSD): $C = C^T$ and $x^TCx \ge 0$ for all $x$.
Because $C$ is SPSD, its eigenvalues are its singular values and $U = V$ in the SVD.

```python
# Toy example: 3 samples, 3 features
D = np.array([[2, 0, 2],
              [4, 1, 4],
              [0, 1, 1]], dtype=float)

n = D.shape[0]
mu = D.mean(axis=0)
Z  = D - mu                  # center
C  = (Z.T @ Z) / n           # covariance matrix

print("C =\n", C)
print("symmetric:", np.allclose(C, C.T))
print("rank:",      npla.matrix_rank(C))   # at most n-1 = 2
print("det:",       npla.det(C))           # 0, because rank < p
```

### Principal components

Compute the SVD of $C$:

>$$C = U \Sigma U^T$$

(using $U=V$ because $C$ is SPSD). The columns $e_i$ of $U$ are the **principal components**.

**Key property:** $Ze_0$ has the *maximum variance* of any unit linear combination of features.
$Ze_1$ has the maximum variance subject to being orthogonal to $e_0$, and so on.

The $i$-th singular value $\sigma_i$ equals that maximum variance:

$$\text{Var}(Ze_i) = \sigma_i, \qquad \frac{\sigma_i}{\sum_j \sigma_j} = \text{fraction of variance explained by PC}_i.$$

```python
U, sigma, Vt = npla.svd(C)

print("Singular values (= eigenvalues of C):", sigma)
print("Variance explained (%):", 100 * sigma / sigma.sum())

# Eigenvectors and singular vectors agree (up to sign)
evals, evecs = npla.eig(C)
order = np.argsort(evals)[::-1]
print("\nEigenvalues sorted:", evals[order].real)
print("Singular values:    ", sigma)
```

### Dimensionality reduction

Project each centered sample onto the first $k$ principal components:

>$$\text{scores} = Z U^{(k)} \in \mathbb{R}^{n \times k}$$

Reconstruct:

$$D^{(k)} \approx M + Z U^{(k)} \left(U^{(k)}\right)^T$$

where $M$ is the matrix with $\mu^T$ stacked $n$ times. Using all $p$ components recovers $D$ exactly.

### Full example: Iris dataset (150 samples, 4 features)

```python
iris = datasets.load_iris()
D, y = iris.data, iris.target          # (150, 4)
n, p = D.shape

mu = D.mean(axis=0)
Z  = D - mu
C  = np.cov(Z.T)                       # (4, 4), uses n-1 denominator (fine here)

U, sigma, _ = npla.svd(C)

print("Variance explained per PC (%):",
      np.round(100 * sigma / sigma.sum(), 2))
```

```python
# Scree plot
cumvar = np.cumsum(sigma) / sigma.sum() * 100
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar([f'PC{i}' for i in range(p)], 100 * sigma / sigma.sum(), label='individual')
ax.plot([f'PC{i}' for i in range(p)], cumvar, 'r-o', label='cumulative')
ax.set_ylabel('% variance')
ax.set_title('Scree plot — Iris')
ax.legend()
plt.tight_layout()
plt.show()
# PC0 alone explains ~92%; PC0+PC1 explain ~98%.
```

```python
# Project onto first two PCs
scores = Z @ U[:, :2]                  # (150, 2)

colors = ['r', 'g', 'b']
labels = iris.target_names

fig, ax = plt.subplots(figsize=(6, 4))
for cls in range(3):
    mask = y == cls
    ax.scatter(scores[mask, 0], scores[mask, 1],
               c=colors[cls], s=15, label=labels[cls])

ax.set_xlabel('PC 0  (92.5% var)')
ax.set_ylabel('PC 1  (5.3% var)')
ax.set_title('Iris projected onto first 2 principal components')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Reconstruction error vs. number of PCs kept
M_mat = np.ones((n, 1)) @ mu.reshape(1, -1)
errs  = []
for k in range(1, p + 1):
    D_k = M_mat + (Z @ U[:, :k]) @ U[:, :k].T
    errs.append(npla.norm(D - D_k, 'fro') / npla.norm(D, 'fro'))

plt.figure(figsize=(4, 3))
plt.plot(range(1, p + 1), errs, '-o')
plt.xlabel('k (PCs kept)')
plt.ylabel('Relative Frobenius error')
plt.title('PCA reconstruction error')
plt.tight_layout()
plt.show()
```

## Part 2 — PageRank

### Setup: from adjacency matrix to Markov matrix

Given a directed graph on $n$ pages with adjacency matrix $E$ ($E_{ij}=1$ means page $j$ links to page $i$):

1. **Handle dangling nodes** — a page with no outgoing links gets uniform links to all others.
2. **Link matrix** $L = E / \text{outdegree}$ — normalize columns to sum to 1.
3. **Markov (Google) matrix** with teleportation probability $m = 0.15$:

>$$M = (1 - m)\,L + m\,\frac{\mathbf{1}\mathbf{1}^T}{n}$$

$M$ is **column-stochastic**: all entries $\ge 0$, columns sum to 1.

### PageRank vector

The PageRank vector $x$ satisfies

>$$M x = x, \qquad \|x\|_1 = 1.$$

$x$ is the dominant eigenvector of $M$ (eigenvalue 1, guaranteed by Perron–Frobenius).
Pages are ranked by the entries of $x$ in descending order.

```python
def make_markov(E, m=0.15):
    """Adjacency matrix E -> column-stochastic Markov matrix M."""
    E = E.astype(float)
    n = E.shape[0]
    outdegree = E.sum(axis=0)
    # fix dangling nodes
    for j in range(n):
        if outdegree[j] == 0:
            E[:, j] = 1.0
            E[j, j] = 0.0
    L = E / E.sum(axis=0)
    return (1 - m) * L + m * np.ones((n, n)) / n
```

```python
# Small example: 4-page graph
#   0 -> 1, 0 -> 2
#   1 -> 2
#   2 -> 0
#   3 -> 0, 3 -> 2
E = np.array([[0, 0, 1, 1],
              [1, 0, 0, 0],
              [1, 1, 0, 1],
              [0, 0, 0, 0]])   # page 3 is a dangling node

M = make_markov(E)
print("M =\n", np.round(M, 3))
print("column sums:", M.sum(axis=0))   # should all be 1
```

### Method 1 — Direct eigenvector

Works for small/medium graphs. Find all eigenvalues, pick the one closest to 1.

```python
d, V = npla.eig(M)
idx  = np.argmax(d.real)           # dominant eigenvalue ≈ 1
x    = V[:, idx].real
x    = np.abs(x) / np.abs(x).sum() # normalize to probability distribution

ranking = np.argsort(x)[::-1]
print("PageRank scores:", np.round(x, 4))
print("Ranking (best first):", ranking)
```

### Method 2 — Power iteration

For large graphs, forming all eigenvectors is too expensive. The power method converges to the dominant eigenvector:

>$$x^{(k+1)} = \frac{M\, x^{(k)}}{\|M\, x^{(k)}\|}, \quad x^{(0)} = \frac{1}{\sqrt{n}}\mathbf{1}.$$

Convergence rate is $|\lambda_2 / \lambda_1|^k$. Because $\lambda_1 = 1$ and $|\lambda_2| \le 1 - m$ for the Google matrix, convergence is geometric with ratio $\le 1 - m = 0.85$.

```python
def pagerank_power(M, tol=1e-10, max_iter=1000):
    n = M.shape[0]
    v = np.ones(n) / np.sqrt(n)
    for i in range(max_iter):
        v_new = M @ v
        v_new /= npla.norm(v_new)
        if npla.norm(v_new - v) < tol:
            print(f'Converged in {i+1} iterations.')
            break
        v = v_new
    x = np.abs(v_new)
    return x / x.sum(), np.argsort(x)[::-1]

x_power, ranking_power = pagerank_power(M)
print("Scores:", np.round(x_power, 4))
print("Ranking:", ranking_power)
```

```python
# Both methods agree
print("Direct vs power agree:", np.allclose(x, x_power, atol=1e-8))
```

### Convergence visualization

```python
n = M.shape[0]
v = np.ones(n) / np.sqrt(n)
x_true = x_power

residuals = []
for _ in range(40):
    v = M @ v
    v /= npla.norm(v)
    residuals.append(npla.norm(np.abs(v) / np.abs(v).sum() - x_true))

plt.figure(figsize=(5, 3))
plt.semilogy(residuals, '-o', markersize=4)
plt.xlabel('Iteration')
plt.ylabel('||x_k - x*||')
plt.title('Power method convergence (log scale)')
plt.tight_layout()
plt.show()
```

### Larger random graph

```python
rng = np.random.default_rng(42)
n   = 50
E_big = (rng.random((n, n)) < 0.1).astype(float)
np.fill_diagonal(E_big, 0)

M_big = make_markov(E_big)
scores, ranking = pagerank_power(M_big)

plt.figure(figsize=(6, 3))
plt.bar(range(n), scores[np.argsort(np.arange(n))])
plt.xlabel('Page')
plt.ylabel('PageRank score')
plt.title(f'PageRank on random 50-node graph  |  top page: {ranking[0]}')
plt.tight_layout()
plt.show()

print('Top 5 pages:', ranking[:5])
print('Their scores:', np.round(scores[ranking[:5]], 4))
```

---
## Summary

| | PCA | PageRank |
|---|---|---|
| **Object** | Covariance matrix $C = \frac{1}{n}Z^TZ$ | Markov matrix $M = (1-m)L + \frac{m}{n}\mathbf{1}\mathbf{1}^T$ |
| **Key property** | SPSD: $\sigma_i = \lambda_i \ge 0$ | Column-stochastic: $\lambda_1 = 1$ |
| **Goal** | Dominant eigenvectors = directions of max variance | Dominant eigenvector = importance scores |
| **Exact method** | `npla.svd(C)` | `npla.eig(M)` |
| **Scalable method** | Truncated SVD (e.g. `svds`) | Power iteration |
| **Convergence rate** | — | Geometric, ratio $\le 1 - m$ |
